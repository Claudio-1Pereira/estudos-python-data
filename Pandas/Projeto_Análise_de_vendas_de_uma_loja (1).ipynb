```python
import pandas as pd
```

```python
dados = pd.read_csv('/content/desafio_pandas_loja.csv')
dados
```

```python
dados['estado'].unique()
```

```python
dados['idade'] = dados['idade'].fillna(dados['idade'].median())
```

```python
dados.dropna(inplace=True)
```

```python
dados.info()
```

```python
dados['data'] = pd.to_datetime(dados['data'])
```

```python
dados['estado'] = dados['estado'].str.strip()
```

```python
dados
```

```python
dados['faturamento'] = dados['quantidade'] * dados['preco_unitario']
dados
```

```python
dados['faturamento'].sum()
```

```python
dados.groupby('produto')['quantidade'].max().sort_values()
```

```python
dados.groupby('produto')['faturamento'].max().sort_values()
```

```python
dados.groupby('categoria')['faturamento'].sum().sort_values()
```

```python
dados.groupby('estado')['faturamento'].sum().sort_values()
```

```python
mes = dados['data'].dt.month
dados.groupby(mes)['faturamento'].sum()
```

```python
dados['idade'].mean()
```

```python
dados['faturamento'].mean()
```

```python
estado = dados.groupby('categoria')['faturamento'].sum().sort_values()
estado.plot(kind='bar', figsize=[10, 10])
```

```python
estado = dados.groupby('estado')['faturamento'].sum().sort_values()
estado.plot(kind='bar', figsize=[10, 10])
```

```python
media = dados.groupby('produto')['faturamento'].sum()
media.plot(kind='bar', figsize=[10, 10])
```
